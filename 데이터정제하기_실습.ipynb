```python
selected_columns = ns_df.columns != 'Unnamed : 13'
ns_book = ns_df.loc[:,selected_columns]
ns_book.head()
```

```python
selected_columns = ns_df.columns != '부가기호'
ns_book = ns_df.loc[:,selected_columns]
ns_book.head()
```

이번에는 데이터프레임의 메서드를 사용하여 열을 삭제
판다스에서는 데이터프레임의 행이나 열을 삭제하는 drop()메서드를 제공 drop()메서드로 열을 삭제하려면 첫번쨰 매개 변수에 삭제하려는 열 이름을 전달하고axis에 1로 지정

```python
ns_book = ns_df.drop(['부가기호','Unnamed: 13'], axis=1)
ns_book.head()
```

지금까지는 ns_df데이터프레임에서 열을 삭제한 결과를 ns_book이라는 변수에 저장 drop()메서드에 inplace매개변수를 True로 지정하면 현재 선택한 데이터프레임을 바로 수정할 수도 있습니다

```python
ns_book.drop('주제분류번호', axis=1, inplace=True)
ns_book.head()
```

새로운 변수를 만들어 데이터프레임 수정결과를 반환하지 않고도 ns_book데이터 프레임에서 '주제분류번호'열을 삭제
판다스는 비어있는 값을 Nan으로 표시합니다
drop()메서드와 비슷한 dropna()메서드는 기본적으로 Nan이 하나 이상ㅎ 포함된 행이나 열을 삭제, axis=1로 지정하여 ns_dfㅍ데이터프레임에서 Nan포함된 열을 삭제 14개의 열이 5개로 확 줄었다 삭제된 열에는 Nandl 적어도 하나 이상 있었나보다

```python
ns_book = ns_df.dropna(axis=1)
ns_book.head()
```

모든 값이 Nan인 열을 삭제하려면 dropna()메서드에 how인 매개변수를 'all'로 지정하면 된다

```python
ns_book = ns_df.dropna(axis=1, how='all')
ns_book.head()
```
